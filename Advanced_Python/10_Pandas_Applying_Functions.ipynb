```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
# Create copy of Data
df_orgl = df.copy()
```

```python
# Use double brackets [[]] so df_salary remains a DataFrame
df_salary = df_orgl [pd.notna(df_orgl['salary_year_avg'])][['salary_year_avg']]
df_salary
```

```python
# Directly calculate the inflated salary column
df_salary['salary_year_inflated'] = df_salary['salary_year_avg']*1.03
# View columns
df_salary[['salary_year_inflated', 'salary_year_avg']]
```

Using Function Apply()

```python
#Calculate projected salaries next year,apply by column.

df_salary = df_orgl[df_orgl['salary_year_avg'].notna()][['salary_year_avg']].copy()

#Define the function 
def projected_salary(salary):
    return salary * 1.03  #using an assumed rate of 3.0% for all roles
# create column salary_year_inflated 
df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(projected_salary)
# View columns
df_salary[['salary_year_inflated', 'salary_year_avg']]
```

Calculate projected salaries next year, but: apply by rows


```python
df_salary = df_orgl [pd.notna(df_orgl['salary_year_avg'])][['salary_year_avg', 'job_title_short']].copy()

def projected_salary(row) : 
    if 'Senior' in row['job_title_short'] : 
      return row['salary_year_avg'] *1.05 ##For senior roles (e.g., Senior Data Analysts), assume the rate is 5%
    else :
      return row['salary_year_avg'] *1.03  #For all other roles, assume rate is 3%
df_salary['salary_year_inflated'] = df_salary.apply(projected_salary, axis =1)
df_salary[['job_title_short','salary_year_inflated', 'salary_year_avg' , ]]
```

```python
# using Apply & Lambda 
df_salary['salary_year_inflated']= df_salary['salary_year_avg'].apply(lambda salary : salary *1.3)
df_salary[['salary_year_inflated', 'salary_year_avg']]
```

```python
df_salary['salary_year_inflated'] = df_salary.apply(lambda row : row['salary_year_avg'] *1.05  if 'Senior' in row['job_title_short'] else row['salary_year_avg'] *1.03 , axis = 1 )

df_salary[['job_title_short','salary_year_inflated', 'salary_year_avg' , ]]
```
