```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

# DataFrame Copy
df_original = df.copy()
```

```python
#Ceate new data frame 
df_altered = df_original 
df_altered.loc[: ,'salary_year_avg']
```

```python
#Calculating the median salary 
median_salary_year = df_altered['salary_year_avg'].median()
#filling the missing values with the median salary 
df_altered['salary_year_avg']= df['salary_year_avg'].fillna(median_salary_year)
```
